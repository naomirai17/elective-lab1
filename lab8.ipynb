```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data= {
    'Category': ['A','B','C','A','B','C','A','B','C','A'],
    'Value': [23,45,56,25,50,52,20,48,55,30]
}

df=pd.DataFrame(data)

sns.boxplot(x='Category', y="Value", data=df)
plt.title("Box Plot of Values by Category")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'Day': ['Mon','Tue','Wed','Thu','Fri','Mon','Tue','Wed','Thu','Fri'],
    'Total_Bill': [20,35,30,35,27,25,32,31,40,28],
    'Gender': ['Male','Male','Female','Female','Male','Female','Male','Male','Female','Female']
}

df = pd.DataFrame(data)

sns.set_style("whitegrid")
plt.figure(figsize=(10,6))

sns.boxplot(
    x='Day',
    y='Total_Bill',
    hue='Gender',
    data=df,
    palette='pastel',
    width=0.5,
    fliersize=8,   # corrected here
    notch=True
)

plt.title("Customized Box Plot of Total Bill by Day and Gender", fontsize=16)
plt.xlabel("Day of the Week", fontsize=12)
plt.ylabel("Total Bill ($)", fontsize=12)
plt.xticks(rotation=30)
sns.despine()
plt.show()
```

```python
# pair plot
data = {
    'Sepal_Length': [5.1,4.9,6.7,5.6,6.3,5.0,5.5,6.0,6.1]
    'Sepal_Width': [3.5,3.0,3.1,2.5,3.3,3.6,2.8,3.0,3.4]
    'Petal_Legth': [1.4,1.4,5.6,4.0,4.8,6.0,1.4,1.5,4.5,5.0]
    'Petal_Width': [0.2,0.2,2.4,1.3,1.8,2.5,0.2,0.2,1.5,1.8]
    'Species': ['Setosa', 'Setosa', 'Vinginica', 'Versicolor','Vinginica','Setosa','Setosa','Versicolor','Vinginica']
}
df= pd.DataFrame(data)
sns.pairplot(df, hue='Species')
```

```python
data= {
    'Math': [90,80,70,85,95],
    'Science': [85,78,75,80,90],
    'English': [88,82,70,78,92],
    'History': [70,65,60,75,80]
}

df= pd.DataFrame(data)

corr= df.corr()

sns.heatmap(
    corr,
    annot=True, #Show correlation values
    fmt=".2f",#Format for annotations
    cmap="coolwarm", #color palette
    linewidths=1, #Lines betwwen cells
    linecolor='white',
    square=True#Make cells square
)
plt.title("Correlation Heat Map of Subjects")
plt.show()
```
