```python
import requests

api_url = "https://jsonplaceholder.typicode.com/todos"
response = requests.get(api_url)

if response.status_code == 200:
    data = response.json()  #list of dictionaries
    print("Data fetched successfully:",len(data))
else:
    print("Failed to fetch data. Status code:", response.status_code)
```

```python
import requests

api_url = "https://jsonplaceholder.typicode.com/todos"

try:
    response = requests.get(api_url,timeout=10) #set timeout for saftety
    if response.status_code == 200:
        data = response.json()  #list of dictionaries
        print("Data fetched successfully:",len(data))
        print("Sample data:",data[:5])  #print first 2 records
    elif response.status_code == 400:
        print("Bad Request: The server could not process the request.")
    elif response.status_code == 401:
        print("Unauthorized: Access is denied due to invalid credentials.")
    elif response.status_code == 403:
        print("Forbidden: You don't have permission to access this resource.")
    elif response.status_code == 404:
        print("Not Found: The requested resource could not be found.")
    else:
        print("Unexpected Error. Status code:", response.status_code)
except requests.exceptions.Timeout:  
    print("The request timed out.")
except requests.exceptions.ConnectionError:
    print("Network error. Please check your connection.")
except requests.exceptions.RequestException as e:
    print("An error occurred:", e)        
```

Import Libraries

Step 1: Import libraries

```python
import requests
from bs4 import BeautifulSoup
```

Step 2: Send HTTP request 

```python
url="https://www.example.com"
response=requests.get(url)
```

Step 3: Check Response Status

```python
if response.status_code==200:
    print("Request Successfully")
else:
    ("Request Failed")
```

Step 4: Parse HTML content

```python
soup=BeautifulSoup(response.content,'html.parser')
```

step 5: View HTML structure

```python
print(soup.prettify())
```

Step 6: Extract Specific Data

```python
links=soup.find_all('a')
for link in links:
    print(link.text.strip(),link.get('href'))

first_link=soup.find('a')
print("First link:",first_link.text.strip(),first_link.get('href'))
```

Step 7: Extract Required Information 

```python
for link in links:
    print(link.text.strip(),link.get('href'))
```

Step 8: Extract Data using Class or ID

```python
heading=soup.find_all('h2')
for h in heading:       
    print(h.text)
```

Step 9: Store Extracted Data

```python
data=[1,2,3]
for h in heading:
    data.append(h.text.strip())
print(data)
```

Step 10: Save Data to CSV

```python
import csv
with open('headings.csv','w',newline='',encoding='utf-8') as f:
    writer=csv.writer(f)
    writer.writerow(['Heading'])
    for item in data:
        writer.writerow([item])
```

Parse HTML

```python
#Parse HTML 
soup=BeautifulSoup(response.content,'html.parser')

#Example 1: find() - first <p> tag
first_p=soup.find('p')
print("Using find():",first_p.text)

#Example 2: find_all() - all <p> tags
all_p=soup.find_all('p')
print("Using find_all():")
for p in all_p:
    print(p.text)

#Example 3: find_all() with class filter
intro_p=soup.find_all('p',class_='intro')
print("Intro paragraphs:") 
for p in intro_p:
    print(p.text)
```
