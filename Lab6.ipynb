```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Sample data
data = {
    'Education': ['High School', 'Bachelors', 'Masters', 'Bachelors', 'Masters']
}

# Create DataFrame
df = pd.DataFrame(data)

# Initialize label encoder
le = LabelEncoder()

# Apply label encoding
df['Education_Encoded'] = le.fit_transform(df['Education'])

print(df)
```

Dummy Values 

```python
#Applying One-Hot Encoding
#One-Hit Encoding coverts categorical columns into **binary columns** so that machine leanrming algorithms can better understand the data.

import pandas as pd

#Sample dataframe
data = {
    'Color': ['Red', 'Blue', 'Green', 'Red', 'Blue']
}
df=pd.DataFrame(data)

#Apply One-Hot Encoding
df_encoded = pd.get_dummies(df, columns=['Color'])
print(df_encoded)
```

```python
import pandas as pd
import numpy as np

#Sample Dataframe
data = {
    'Value': [1, 10, 100, 1000]
}
df = pd.DataFrame(data)

#Apply Log10 Transformation
df['Log_Value'] = np.log(df['Value'])
print(df)
```

```python
import pandas as pd
import numpy as np

#Sample Dataframe
data = {
    'Value': [1, 10, 100, 1000]
}
df = pd.DataFrame(data)

#Apply square root Transformation
df['Sqrt_Value'] = np.sqrt(df['Value'])

print(df)
```

Min Max Scaling

```python
import pandas as pd
import numpy as np

#Sample Dataframe
data = {
    'Score': [50, 100, 150]
}
df = pd.DataFrame(data)

#Min-Max Scaling manually
df['Score_Scaled'] = (df['Score'] - df['Score'].min()) / (df['Score'].max() - df['Score'].min())
print(df)
```

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd
#Create a DataFrame
data = pd.DataFrame({'Value' : [50, 100, 150]})

#Initialize StandardScaler
Scaler = StandardScaler()

#fit and transform the data
data['Value_Standardized'] = Scaler.fit_transform(data[['Value']])
print(data)
```
